## 1. Load a Transformer

We use **DistilBERT**, a  Transformer fine-tuned for positive/negative sentiment classification.

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch

model_name = "distilbert-base-uncased-finetuned-sst-2-english"

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForSequenceClassification.from_pretrained(model_name)
model.eval()

# Look inside the Transformer architecture.
print("Transformer layers:", model.config.n_layers)
print("Attention heads per layer:", model.config.n_heads)
print("Hidden size:", model.config.dim)
```


<img src="https://drive.google.com/thumbnail?id=1v2dRK1KR_ydZOknjj8Xv8_wjOdXleDYs&sz=w1000" width="500">

## 2. Tokenizing the text

```python
sentence = "The movie was not good."

# Tokenization: converts text into token IDs
inputs = tokenizer(sentence, return_tensors="pt")

# [CLS] = start token, [SEP] = end token
tokens = tokenizer.convert_ids_to_tokens(inputs["input_ids"][0])

print("Tokens:", tokens)
```

## 3. Embeddings and self-attention

Each token ID becomes a **vector (embedding)**. Inside each Transformer layer, **self-attention** lets a token use information from other tokens in the same sentence.

```python
import torch

# Enable attention weights
model.set_attn_implementation("eager")

with torch.no_grad():
    embeddings = model.distilbert.embeddings(inputs["input_ids"])
    outputs = model(**inputs, output_attentions=True)

# Attention from the first layer and first head
attention = outputs.attentions[0][0, 0]

print("Embeddings shape:", embeddings.shape)
print("Attention shape:", attention.shape)
print("Tokens:", tokens)
print("Attention weights:\n", attention)
```

## 4. Classification: From Transformer to prediction

The Transformer produces contextual representations. The **classification head** turns them into two scores (*logits*). **Softmax** converts these into probabilities that sum to 1.

```python
# Get prediction scores from the Transformer
with torch.no_grad():
    logits = model(**inputs).logits

# Convert scores into probabilities
probs = torch.softmax(logits, dim=1)[0]

# Display the prediction
prediction = probs.argmax().item()

print("Prediction:", model.config.id2label[prediction])
print("Confidence:", round(probs[prediction].item(), 3))
```

### Task Solution

The following cells load `bert-base-uncased`, inspect its architecture, and extract attention from the **third Transformer layer** and **second attention head**. Python uses zero-based indexing, so these correspond to layer index `2` and head index `1`.


```python
from transformers import AutoTokenizer, AutoModel
import torch
import matplotlib.pyplot as plt

# Load the pretrained BERT model and tokenizer
bert_model_name = "bert-base-uncased"
bert_tokenizer = AutoTokenizer.from_pretrained(bert_model_name)
bert_model = AutoModel.from_pretrained(
    bert_model_name,
    attn_implementation="eager"
)
bert_model.eval()

# 1. Number of Transformer layers and attention heads per layer
num_layers = bert_model.config.num_hidden_layers
num_heads = bert_model.config.num_attention_heads

print("Number of Transformer layers:", num_layers)
print("Attention heads per layer:", num_heads)

# 2. Total number of attention heads across all layers
total_heads = num_layers * num_heads
print("Total number of attention heads:", total_heads)

# 3. Extract attention from the third layer and second head
sentence = "Transformers use self-attention to understand language."
bert_inputs = bert_tokenizer(sentence, return_tensors="pt")
bert_tokens = bert_tokenizer.convert_ids_to_tokens(bert_inputs["input_ids"][0])

with torch.no_grad():
    bert_outputs = bert_model(**bert_inputs, output_attentions=True)

# attentions: tuple of layers; each tensor has shape:
# (batch_size, number_of_heads, sequence_length, sequence_length)
third_layer_attention = bert_outputs.attentions[2]
second_head_attention = third_layer_attention[0, 1]

print("\nTokens:", bert_tokens)
print("Attention matrix shape:", tuple(second_head_attention.shape))
print("Attention weights (third layer, second head):")
print(second_head_attention)

# Optional visualization of the extracted attention matrix
attention_matrix = second_head_attention.cpu().numpy()
plt.figure(figsize=(10, 8))
plt.imshow(attention_matrix, cmap="viridis", aspect="auto")
plt.xticks(range(len(bert_tokens)), bert_tokens, rotation=45, ha="right")
plt.yticks(range(len(bert_tokens)), bert_tokens)
plt.xlabel("Key tokens (columns)")
plt.ylabel("Query tokens (rows)")
plt.title("BERT: Third Layer, Second Attention Head")
plt.colorbar(label="Attention weight")
plt.tight_layout()
plt.show()
```

### 4. Explanation of the attention matrix

- **Rows (query tokens):** each row represents the token whose representation is being updated. It shows how much that token attends to every token in the input sequence.
- **Columns (key tokens):** each column represents a token that the row token can attend to and gather information from.
- **Each cell:** the value at row *i*, column *j* is the attention weight from query token *i* to key token *j*. Larger values indicate stronger attention.
- **Each row sums to approximately 1** because attention weights are produced by a softmax operation.

For BERT-base-uncased, the expected architecture values are **12 Transformer layers**, **12 attention heads per layer**, and **144 attention heads in total**. The exact attention weights depend on the pretrained model and are generated when the code is run.
